```python
import cv2
import matplotlib.pyplot as plt
```

```python
image_path = "flower.jpg"
gray_image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
```

```python
plt.figure(figsize=(5,5))
plt.imshow(gray_image, cmap='gray')
plt.title("Original Grayscale Image")
plt.axis('off')
plt.show()
```

```python
scales = [2.0, 4.0, 0.5, 0.25]
```

```python
plt.figure(figsize=(12, 10))

for i, scale in enumerate(scales):

    resized_image = cv2.resize(gray_image, None, fx=scale, fy=scale, interpolation=cv2.INTER_NEAREST)  
    
    plt.subplot(2, 2, i + 1)
    plt.imshow(resized_image, cmap='gray')
    plt.title(f"Scale {scale}")
    plt.axis('off')

plt.suptitle("INTER_NEAREST Interpolation", fontsize=16)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 10))

for i, scale in enumerate(scales):

    resized_image = cv2.resize(gray_image, None, fx=scale, fy=scale, interpolation=cv2.INTER_LINEAR)  

    plt.subplot(2, 2, i + 1)
    plt.imshow(resized_image, cmap='gray')
    plt.title(f"Scale {scale}")
    plt.axis('off')

plt.suptitle("INTER_LINEAR Interpolation", fontsize=16)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 10))

for i, scale in enumerate(scales):

    resized_image = cv2.resize(gray_image, None, fx=scale, fy=scale, interpolation=cv2.INTER_CUBIC)

    plt.subplot(2, 2, i + 1)
    plt.imshow(resized_image, cmap='gray')
    plt.title(f"Scale {scale}")
    plt.axis('off')

plt.suptitle("INTER_CUBIC Interpolation", fontsize=16)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 10))

for i, scale in enumerate(scales):

    resized_image = cv2.resize(gray_image, None, fx=scale, fy=scale, interpolation=cv2.INTER_LANCZOS4)
    
    plt.subplot(2, 2, i + 1)
    plt.imshow(resized_image, cmap='gray')
    plt.title(f"Scale {scale}")
    plt.axis('off')

plt.suptitle("INTER_LANCZOS4 Interpolation", fontsize=16)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 10))

for i, scale in enumerate(scales):

    resized_image = cv2.resize(gray_image, None, fx=scale, fy=scale, interpolation=cv2.INTER_AREA)
    
    plt.subplot(2, 2, i + 1)
    plt.imshow(resized_image, cmap='gray')
    plt.title(f"Scale {scale}")
    plt.axis('off')

plt.suptitle("INTER_AREA Interpolation", fontsize=16)

plt.tight_layout()
plt.show()
```

```python
interpolation_methods = {"INTER_NEAREST": cv2.INTER_NEAREST,
                         "INTER_LINEAR": cv2.INTER_LINEAR,
                         "INTER_CUBIC": cv2.INTER_CUBIC,
                         "INTER_LANCZOS4": cv2.INTER_LANCZOS4,
                         "INTER_AREA": cv2.INTER_AREA}
```

```python
plt.figure(figsize=(15, 12))

plot_number = 1

for method_name, method in interpolation_methods.items():

    for scale in scales:

        resized_image = cv2.resize(gray_image, None, fx=scale, fy=scale, interpolation=method)

        plt.subplot(len(interpolation_methods), len(scales), plot_number)
        plt.imshow(resized_image, cmap='gray')
        plt.title(f"{method_name}\nScale {scale}")
        plt.axis('off')

        plot_number += 1

plt.tight_layout()
plt.show()
```
